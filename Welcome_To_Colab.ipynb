```
import sqlite3

class BusReservation:

    def __init__(self):
        self.conn = sqlite3.connect("bus_reservation.db")
        self.cur = self.conn.cursor()

        self.cur.execute('''
        CREATE TABLE IF NOT EXISTS buses(
            bus_no TEXT PRIMARY KEY,
            source TEXT,
            destination TEXT,
            seats INTEGER
        )
        ''')

        self.cur.execute('''
        CREATE TABLE IF NOT EXISTS bookings(
            booking_id INTEGER PRIMARY KEY AUTOINCREMENT,
            passenger_name TEXT,
            bus_no TEXT,
            seat_no INTEGER
        )
        ''')

        self.conn.commit()

    def add_bus(self):
        bus_no = input("Enter Bus Number: ")
        source = input("Enter Source: ")
        destination = input("Enter Destination: ")
        seats = int(input("Enter Total Seats: "))

        self.cur.execute(
            "INSERT INTO buses VALUES(?,?,?,?)",
            (bus_no, source, destination, seats)
        )
        self.conn.commit()

        print("Bus Added Successfully!")

    def view_buses(self):
        self.cur.execute("SELECT * FROM buses")
        buses = self.cur.fetchall()

        print("\nAvailable Buses")
        print("-"*50)

        for bus in buses:
            print(f"Bus No: {bus[0]}")
            print(f"Source: {bus[1]}")
            print(f"Destination: {bus[2]}")
            print(f"Seats: {bus[3]}")
            print("-"*50)

    def book_ticket(self):
        name = input("Passenger Name: ")
        bus_no = input("Bus Number: ")
        seat_no = int(input("Seat Number: "))

        self.cur.execute(
            "SELECT * FROM bookings WHERE bus_no=? AND seat_no=?",
            (bus_no, seat_no)
        )

        if self.cur.fetchone():
            print("Seat Already Booked!")
            return

        self.cur.execute(
            "INSERT INTO bookings(passenger_name,bus_no,seat_no) VALUES(?,?,?)",
            (name, bus_no, seat_no)
        )

        self.conn.commit()

        print("Ticket Booked Successfully!")

    def cancel_ticket(self):
        booking_id = int(input("Enter Booking ID: "))

        self.cur.execute(
            "DELETE FROM bookings WHERE booking_id=?",
            (booking_id,)
        )

        self.conn.commit()

        print("Ticket Cancelled Successfully!")

    def view_bookings(self):
        self.cur.execute("SELECT * FROM bookings")
        data = self.cur.fetchall()

        print("\nBooking Details")
        print("-"*60)

        for row in data:
            print(f"Booking ID : {row[0]}")
            print(f"Passenger  : {row[1]}")
            print(f"Bus Number : {row[2]}")
            print(f"Seat No    : {row[3]}")
            print("-"*60)

    def close(self):
        self.conn.close()


system = BusReservation()

while True:
    print("\n===== BUS TICKET RESERVATION SYSTEM =====")
    print("1. Add Bus")
    print("2. View Buses")
    print("3. Book Ticket")
    print("4. Cancel Ticket")
    print("5. View Bookings")
    print("6. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":
        system.add_bus()

    elif choice == "2":
        system.view_buses()

    elif choice == "3":
        system.book_ticket()

    elif choice == "4":
        system.cancel_ticket()

    elif choice == "5":
        system.view_bookings()

    elif choice == "6":
        system.close()
        print("Thank You!")
        break

    else:
        print("Invalid Choice!")
```
